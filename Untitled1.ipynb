```python
import datetime
import re


def get_response(user_input):
    # Convert input to lowercase and remove leading/trailing whitespace
    text = user_input.lower().strip()

    # 1. Greetings
    if re.search(r"\b(hi|hello|hey|greetings)\b", text):
        return "Hello! How can I assist you today?"

    # 2. Asking how the bot is doing
    elif re.search(r"\b(how are you|how do you do|how's it going)\b", text):
        return "I'm just a computer program, but I'm functioning smoothly! How are you?"

    # 3. Asking for bot's identity/name
    elif re.search(r"\b(what is your name|who are you)\b", text):
        return "I am a simple rule-based AI chatbot created for the CodSoft internship!"

    # 4. Inquiring about current time
    elif re.search(r"\b(time|what time is it)\b", text):
        now = datetime.datetime.now()
        return f"The current system time is {now.strftime('%I:%M %p')}."

    # 5. Inquiring about current date
    elif re.search(r"\b(date|what day is today)\b", text):
        today = datetime.date.today()
        return f"Today's date is {today.strftime('%B %d, %Y')}."

    # 6. Help / Capabilities
    elif re.search(r"\b(help|what can you do)\b", text):
        return (
            "I can answer basic greetings, tell you the time/date, "
            "answer simple questions, or just chat! Type 'bye' to exit."
        )

    # 7. Gratitude
    elif re.search(r"\b(thank you|thanks)\b", text):
        return "You're very welcome! Let me know if you need anything else."

    # 8. Goodbyes
    elif re.search(r"\b(bye|goodbye|exit|quit)\b", text):
        return "Goodbye! Have a productive day ahead!"

    # Fallback response if no rules match
    else:
        return "I'm sorry, I didn't quite catch that. Could you try rephrasing or type 'help'?"


def run_chatbot():
    print("=" * 50)
    print("      RULE-BASED AI CHATBOT (Type 'bye' to exit)     ")
    print("=" * 50)

    while True:
        user_input = input("You: ")

        # Check if the user wants to leave
        if user_input.lower().strip() in ["bye", "exit", "quit", "goodbye"]:
            print(f"Bot: {get_response(user_input)}")
            break

        response = get_response(user_input)
        print(f"Bot: {response}\n")


if __name__ == "__main__":
    run_chatbot()
```
