```python
import datetime
import random
import re


def get_response(user_input):
    text = user_input.lower().strip()

    # 1. Greetings
    if re.search(r"\b(hi|hello|hey|greetings|good morning|good evening)\b", text):
        return "Hello! How can I assist you today?"

    # 2. Status / Wellbeing
    elif re.search(r"\b(how are you|how do you do|how are you doing)\b", text):
        return "I'm running smoothly, thank you! How can I help you today?"

    # 3. Bot Identity
    elif re.search(r"\b(what is your name|who are you)\b", text):
        return "I am an AI Rule-Based Chatbot built for the CodSoft Internship!"

    # 4. Time
    elif re.search(r"\b(time|what time is it)\b", text):
        return (
            f"The current time is {datetime.datetime.now().strftime('%I:%M %p')}."
        )

    # 5. Date / Day
    elif re.search(r"\b(date|today|what day is it)\b", text):
        return f"Today's date is {datetime.date.today().strftime('%B %d, %Y')}."

    # 6. Questions about AI / Machine Learning
    elif re.search(
        r"\b(what is ai|artificial intelligence|machine learning)\b", text
    ):
        return (
            "Artificial Intelligence (AI) enables machines to simulate human intelligence, "
            "such as problem-solving, learning, and decision-making!"
        )

    # 7. Questions about Python
    elif re.search(r"\b(what is python|python)\b", text):
        return "Python is a popular, beginner-friendly programming language widely used in AI, data science, and web development."

    # 8. Weather
    elif re.search(r"\b(weather|is it raining|temperature)\b", text):
        return "I don't have real-time GPS access, but you can check your local forecast app for the latest weather updates!"

    # 9. Jokes
    elif re.search(r"\b(joke|tell me a joke|make me laugh)\b", text):
        jokes = [
            "Why do programmers prefer dark mode? Because light attracts bugs!",
            "Why did the computer show up at work late? It had a hard drive!",
            "There are only 10 types of people in the world: those who understand binary and those who don't.",
        ]
        return random.choice(jokes)

    # 10. Help / Commands
    elif re.search(r"\b(help|commands|what can you do)\b", text):
        return (
            "You can ask me about:\n"
            "- Greetings ('hello')\n"
            "- Time or Date ('what time is it?')\n"
            "- Topics like 'what is AI' or 'what is Python'\n"
            "- Ask for a 'joke'\n"
            "- Type 'bye' to exit."
        )

    # 11. Gratitude
    elif re.search(r"\b(thank you|thanks)\b", text):
        return "You're very welcome! Let me know if you need anything else."

    # 12. Goodbyes
    elif re.search(r"\b(bye|exit|quit|goodbye)\b", text):
        return "Goodbye! Have a great day ahead!"

    # Fallback
    else:
        return "I'm not sure how to answer that yet. Type 'help' to see what topics I know!"


def run_chatbot():
    print("=" * 55)
    print("      RULE-BASED AI CHATBOT (Type 'bye' to exit)     ")
    print("=" * 55)

    while True:
        user_input = input("You: ")
        if not user_input.strip():
            continue

        if user_input.lower().strip() in ["bye", "exit", "quit", "goodbye"]:
            print(f"Bot: {get_response(user_input)}")
            break

        print(f"Bot: {get_response(user_input)}\n")


if __name__ == "__main__":
    run_chatbot()
```
