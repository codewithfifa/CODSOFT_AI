```python
import math

# Board representation: 9 empty spaces
board = [" " for _ in range(9)]


def print_board():
    """Prints the current state of the board in a 3x3 grid."""
    print()
    for row in [board[i * 3 : (i + 1) * 3] for i in range(3)]:
        print("| " + " | ".join(row) + " |")
    print()


def available_moves():
    """Returns a list of available indices (0-8)."""
    return [i for i, spot in enumerate(board) if spot == " "]


def check_winner(b, player):
    """Checks if the given player has won the game."""
    win_conditions = [
        # Horizontal
        [0, 1, 2],
        [3, 4, 5],
        [6, 7, 8],
        # Vertical
        [0, 3, 6],
        [1, 4, 7],
        [2, 5, 8],
        # Diagonal
        [0, 4, 8],
        [2, 4, 6],
    ]
    return any(all(b[pos] == player for pos in line) for line in win_conditions)


def is_board_full():
    """Checks if there are no spots left."""
    return " " not in board


# ------------------- MINIMAX ALGORITHM -------------------
def minimax(state, is_maximizing):
    """Minimax evaluation function for optimal AI move."""
    # Terminal states
    if check_winner(state, "O"):  # AI wins
        return {"score": 1}
    elif check_winner(state, "X"):  # Human wins
        return {"score": -1}
    elif " " not in state:  # Tie
        return {"score": 0}

    if is_maximizing:
        best = {"score": -math.inf, "position": None}
        for move in [i for i, s in enumerate(state) if s == " "]:
            state[move] = "O"
            sim_score = minimax(state, False)
            state[move] = " "  # Undo move
            sim_score["position"] = move

            if sim_score["score"] > best["score"]:
                best = sim_score
        return best
    else:
        best = {"score": math.inf, "position": None}
        for move in [i for i, s in enumerate(state) if s == " "]:
            state[move] = "X"
            sim_score = minimax(state, True)
            state[move] = " "  # Undo move
            sim_score["position"] = move

            if sim_score["score"] < best["score"]:
                best = sim_score
        return best


# ------------------- GAMEPLAY LOOP -------------------
def play_game():
    print("=" * 45)
    print("       TIC-TAC-TOE AI (MINIMAX AGENT)        ")
    print("=" * 45)
    print("Positions are numbered 0 to 8 like this:")
    print("| 0 | 1 | 2 |\n| 3 | 4 | 5 |\n| 6 | 7 | 8 |")
    print("-" * 45)
    print("You are 'X' and AI is 'O'. You go first!")

    print_board()

    while True:
        # --- HUMAN TURN (X) ---
        while True:
            try:
                user_choice = int(input("Enter your move (0-8): "))
                if user_choice in available_moves():
                    board[user_choice] = "X"
                    break
                else:
                    print("That position is already taken or invalid. Try again.")
            except ValueError:
                print("Please enter a valid number between 0 and 8.")

        print("\nYour Move:")
        print_board()

        if check_winner(board, "X"):
            print("🎉 Congratulations! You won!")
            break
        if is_board_full():
            print("🤝 It's a tie!")
            break

        # --- AI TURN (O) ---
        print("AI is thinking...")
        best_move = minimax(board, True)["position"]
        board[best_move] = "O"

        print(f"\nAI chose position {best_move}:")
        print_board()

        if check_winner(board, "O"):
            print("🤖 AI wins! Better luck next time.")
            break
        if is_board_full():
            print("🤝 It's a tie!")
            break


if __name__ == "__main__":
    play_game()
```
